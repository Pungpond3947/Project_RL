```python
ep0 = df[df["episode_id"] == 0]
```

```python
plt.figure()
plt.plot(ep0["timestep"], ep0["robot_pos_x"])
plt.xlabel("Timestep")
plt.ylabel("Robot X Position")
plt.title("Robot Position Over Time")
plt.show()

plt.figure()
plt.plot(ep0["timestep"], ep0["total_reward"])
plt.xlabel("Timestep")
plt.ylabel("Total Reward")
plt.title("Reward Over Time")
plt.show()

plt.figure()
plt.plot(ep0["timestep"], ep0["pos0"], label="Joint0")
plt.plot(ep0["timestep"], ep0["pos1"], label="Joint1")
plt.plot(ep0["timestep"], ep0["pos2"], label="Joint2")
plt.legend()
plt.title("Joint Position (Normalized)")
plt.show()


plt.figure()
plt.plot(ep0["timestep"], ep0["vel0"], label="Joint0")
plt.plot(ep0["timestep"], ep0["vel1"], label="Joint1")
plt.plot(ep0["timestep"], ep0["vel2"], label="Joint2")
plt.legend()
plt.title("Joint Velocity (Normalized)")
plt.show()

plt.figure()
plt.plot(ep0["timestep"], ep0["a0"], label="a0")
plt.plot(ep0["timestep"], ep0["a1"], label="a1")
plt.plot(ep0["timestep"], ep0["a2"], label="a2")
plt.legend()
plt.title("Action Output")
plt.show()

plt.figure()
plt.plot(ep0["timestep"], ep0["a0"], label="a0")
plt.plot(ep0["timestep"], ep0["a1"], label="a1")
plt.plot(ep0["timestep"], ep0["a2"], label="a2")
plt.legend()
plt.title("Action Output")
plt.show()
```

```python
plt.figure()

plt.plot(ep0["timestep"], ep0["foot_pos_x"], label="foot_x")
plt.plot(ep0["timestep"], ep0["foot_pos_y"], label="foot_y")
plt.plot(ep0["timestep"], ep0["foot_pos_z"], label="foot_z")

plt.legend()
plt.title("End Effector Position")
plt.xlabel("Timestep")
plt.ylabel("Position (m)")
plt.show()
```

```python
plt.figure()

plt.plot(ep0["foot_pos_y"], ep0["foot_pos_z"])

plt.xlabel("Foot Y (forward)")
plt.ylabel("Foot Z (height)")
plt.title("Foot Trajectory (X-Z plane)")
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot(ep0["foot_pos_y"], ep0["foot_pos_x"], ep0["foot_pos_z"])

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_title("End Effector 3D Trajectory")

plt.show()
```

```python
plt.figure()

plt.scatter(ep0["phase"], ep0["foot_pos_z"], s=3)

plt.xlabel("Phase")
plt.ylabel("Foot height")
plt.title("Foot height vs phase")

plt.show()
```
